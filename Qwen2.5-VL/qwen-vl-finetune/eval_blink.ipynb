```python
import json
import numpy as np
import os
import re
## open ground_truths

blink3 = []
with open("/mmfs1/gscratch/krishna/mahtab/Aurora-perception/Data/evals/hardblink/answers/blink_3pointscenter_answers.jsonl", "r", encoding="utf-8") as file:
    for line in file:
        blink3.append(json.loads(line.strip()))

blink4 = []
with open("/mmfs1/gscratch/krishna/mahtab/Aurora-perception/Data/evals/hardblink/answers/blink_4pointscenter_answers.jsonl", "r", encoding="utf-8") as file:
    for line in file:
        blink4.append(json.loads(line.strip()))

blink5 = []
with open("/mmfs1/gscratch/krishna/mahtab/Aurora-perception/Data/evals/hardblink/answers/blink_5pointscenter_answers.jsonl", "r", encoding="utf-8") as file:
    for line in file:
        blink5.append(json.loads(line.strip()))



```

```python
def calc_acc_short(pred_data, gt_data):
    acc, total = 0,0
    for k,pred in pred_data.items():
        gt = gt_data[int(k)]['text']
        if not (len(gt) == 3 and len(pred)==3 and gt[1] in ['A','B','C','D','E']):
            print(pred)
            pred =  input()
        if gt.lower() == pred.lower():
            acc+=1
    
        total +=1
        # print(gt,pred)
    # print(acc, total)
    return acc/total*100.0
```

```python

## blink 3
answers = {}
OUTPUT_DIR = "/mmfs1/gscratch/krishna/mahtab/mmseek/Qwen2.5-VL/qwen-vl-finetune/outputs/"
short_answer_files = [f for f in os.listdir(OUTPUT_DIR) if "short" in f and "blink3" in f and "Qwen" not in f] 
for pred_file in short_answer_files:
    with open(f"/mmfs1/gscratch/krishna/mahtab/mmseek/Qwen2.5-VL/qwen-vl-finetune/outputs/{pred_file}", "r") as f:
        pred_data = json.load(f)
    acc = calc_acc_short(pred_data, blink3)
    answers[pred_file.replace("_short.json", "")] = acc
answers    
```

```python
def calc_acc_long(pred_data, gt_data):
    acc, total = 0,0
    for k,v in pred_data.items():
        gt = gt_data[int(k)]['text']
        # print(gt,v)
        m = re.search(r"Since point (.) has a higher pixel value on the depth map", v)
        if m is not None:
            pred = m.group(1)
            if not (len(gt) == 3 and len(pred)==1 and gt[1] in ['A','B','C','D','E']):
                pred =  input()
            if gt[1].lower() == pred.lower():
                acc+=1
            total +=1
            # print(gt[1],pred)
    # print(acc, total)
    return acc/total*100.0, total
```

```python
def calc_acc_qwen3(pred_data, gt_data):
    acc, total = 0,0
    for k,v in pred_data.items():
        gt = gt_data[int(k)]['text']
        v = v.replace("\"", "")
        v = v.replace("\'", "")
        try:
            pred = re.search(r"(.) is the closest to the camera", v).group(1)
        except:
            print(v)
        if  pred not in ['A','B','C','D','E']:
            print(v)
            pred = input()
        if gt[1].lower() == pred.lower():
            acc+=1
        total +=1
        # print(gt,pred)
    # print(acc, total)
    return acc/total*100.0
```
