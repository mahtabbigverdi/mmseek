```python
import json
with open("/mmfs1/gscratch/krishna/mahtab/LLaVA/train_baselinewithcoords90k_annealing_data.json", "r") as f:
    data = json.load(f)

len(data)

data[0]
```

```python
new_data =[]
for i in range(len(data)):
    example = data[i]['conversations']
    new_data.append({
    "conversations": [
      {
        "from": "human",
        "value": example[0]['value']
      },
      {
        "from": "gpt",
        "value": example[1]['value']
      }
    ],
    "images": [data[i]['image']],
  })
len(new_data)   
```

```python
with open('/mmfs1/gscratch/krishna/mahtab/mmseek/LLaMA-Factory/data/baseline_annealing.json', 'w') as f:
    json.dump(new_data, f, ensure_ascii=False, indent=2)
```
